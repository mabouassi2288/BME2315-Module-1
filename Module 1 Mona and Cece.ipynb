## Disease Background: 

* **Prevalence & incidence:** In the United States, Alzheimer’s disease’s greatest risk factor is increasing age, leading to a high prevalence among senior citizens.  Currently, approximately 7.4 million Americans aged 65 years and older have been diagnosed with Alzheimer’s disease, yielding a prevalence of 12.09%. Case reports from 2011 indicate about 910,000 cases of Alzheimer’s out of 41.1 million Americans aged 65 plus, making the annual incidence rate estimate roughly 2.2%. Between 2026 and 2060, researchers project that Alzheimer’s cases will surge by 6.4 million to reach 13.8 million by 2060, with an incidence of 6.7%.  Additionally, early-onset dementia impacts 200,000 individuals ages 30 to 64 with a 0.13% prevalence in this age population.

* **Economic burden:** Alzheimer’s disease is a prolonged condition that strips individuals of their identity and reduces their quality of life placing a heavy mental, physical and economic toll on caregivers. The total U.S cost of Alzheimer’s disease is approximately $781 billion making it one of the most expensive conditions in the country. Direct costs($232 billion) cover medical care, nursing homes, assisted living and home modifications. Indirect costs ($233 billion) include unpaid caregiving and reduced quality of life. These expenses lead to a heavy cost burden on the family and often lead to financial strain, unpaid leave, and job losses, resulting in over $400 billion in broader societal costs.
Source: https://www.healthaffairs.org/sponsored-content/lilly/the-hidden-economic-costs-of-alzheimers-disease 

* **Risk factors (genetic, lifestyle):** 
    - Genetic/Biological: Underlying conditions can increase the likeliness of a future Alzheimer’s diagnosis. Individuals with down syndrome produce three copies of chromosome 21, a gene involved with the production of proteins that create Amyloid-beta. These protein fragments create plagues between synapses in neurotransmitters within the brain, heavily disrupting neuron communication. Other conditions such as hearing and vision loss can impact the likelihood of Alzheimer development if gone untreated due to its reliance on overworking the brain and increasing cognitive decline.
    - Modifiable lifestyle: unhealthy behaviors: Unhealthy lifestyle choices and substance abuse directly correlate with Alzheimer’s disease risk.  Smoking, drinking, and substance abuse can increase the risk of developing Alzheimer’s. Eliminating these behaviors and substituting new lifestyle changes such as adopting a new diet with fresh produce, less cholesterol, or changing to a Mediterranean diet can help reduce risk. Additionally, staying engaged mentally through socializing and physical activity can help stimulate brain activity and help with neural connections.
    - Source: https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/symptoms-causes/syc-20350447 


* **Societal determinants:** Social determinants are the non-medical factors in environments where people live, work, and learn to shape the risk of developing Alzheimer’s. Disadvantaged environments often limit education levels and studies show an association between low education and poor brain health is more prevalent in individuals without a high school diploma. In neighborhoods with accessible healthcare, people are more likely to avoid disease by identifying risky behaviors and conditions to receive early treatment. Without a good healthcare system, people often have lower optimal health in general. Lastly, living in isolation reduces social contact negatively impacting brain cognition. Ultimately, these conditions are heavily influenced by the society that a person grows up in.
Source: https://www.cdc.gov/alzheimers-dementia/php/sdoh/index.html 

* **Symptoms:** Memory loss is the primary symptom of Alzheimer's disease, showing up early on, and progressively worsening. Memory loss can also be associated with thinking and clear reasoning. As the disease progresses, many individuals find it difficult to focus on tasks, make poor decisions, and say repeated statements. These symptoms also pose a strong danger to individuals, especially if living alone. For example, a senior attempting to cook scrambled eggs may include the eggshells without realizing. Alzheimer’s also has the capability to alter behavior causing depression, mood swings, anger, depression, and loss of inhibitions. Without a diagnosis, family members may mistake this for unruliness and violence. Yet, despite the following symptoms, some people with Alzheimer’s disease may keep their “preserved skills,” ex. reading, dancing and crafts.
Source: https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/symptoms-causes/syc-20350447 

* **Diagnosis:** 
    - Recognizing early symptoms is the first step towards receiving an Alzheimer’s disease diagnosis. Seeking medical help for a loved one can be exceptionally challenging if they exhibit symptoms such as aggression, anger or denial. Overcoming this obstacle is essential, as the most effective treatment comes with an early diagnosis.
    - Primary testing is done through biomarkers, imaging and neurophysiological evaluations.
        - Positron Emission Tomography (PET) scans detect the accumulation of plague and tangles caused by amyloid-beta and tau proteins in the brain tissue that disrupt the signaling between neurotransmitters.  Assessing these biomarkers helps neurologists to understand how protein accumulation correlates to cognitive decline.
        - Neurophysiological and mental testing measures cognitive skills and memory, while physical tests are also used to evaluate motor functions, reflexes and balance.
        - Genetic testing can be considered for individuals with a family history with Alzheimer’s disease.
    - Following a diagnosis, the next steps focus on developing a care plan, evaluating treatment options, and making in-home safety modifications. While an early diagnosis provides more effective care, implementing the right lifestyle and social changes helps ease the decline in quality of life.
    - Source: https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/diagnosis-treatment/drc-20350453 


* **Standard of care treatments (& reimbursement):** There is no cure for Alzheimer’s disease, but there are a multitude of treatments to address the disease’s symptoms. Cholinesterase inhibitors are given to patients to reduce or control cognitive and behavioral symptoms. These work by preventing the breakdown of acetylcholine, which is believed to be vital to thinking and memory. Lecanemab and donanemab are medications given to reduce amyloid plaque, which has shown in studies to slow the rate of cognitive decline. Memantine is prescribed to sustain Alzheimer’s patients a little longer and help them maintain certain daily functions. It regulates glutamate, which can lead to brain cell death if too much is produced. 

As for reimbursement, there are blood-based biomarkers covered by Medicare that have a reimbursement pathway of a certain amount per validated plasma ratio test. Medicare Part B covers PET scanning, monoclonal antibodies (Leqembi, Kisunla), and cognitive care planning.

* **Disease progression & prognosis:** There are five stages associated with Alzheimer’s disease: preclinical Alzheimer’s disease, mild cognitive impairment, mild dementia, moderate dementia, and severe dementia due to Alzheimer’s disease. Preclinical Alzheimer’s disease begins before physical and mental symptoms arise, and it consists of amyloid plaques and pTAU tangles developing in the brain. These are biomarkers of Alzheimer’s disease and can be found with new technology. Mild cognitive impairment occurs when there are mild changes in thinking ability and memory, but not significant enough to affect relationships or work. Mild dementia is when patients may experience memory loss of recent events, changes in personality, trouble with problem-solving/organizing, and getting lost/misplacing belongings. Moderate dementia is when the patient has increasingly poor judgement, confusion, memory loss, and inability to perform daily activities. The final stage, severe dementia, occurs when the patient loses the ability to communicate, independently complete daily activities, and potentially the ability to walk without assistance. The rate of progression from the first to the last stage is typically between three and eleven years, but it can be more than twenty years for some. 
Source: https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/in-depth/alzheimers-stages/art-20048448

* **Continuum of care providers:** In the early stages of Alzheimer’s disease, patients experience home care, where there are safety modifications at home and a family support system. The next stage is in an assisted living facility or adult daycare, which provides help with eating, bathing, and other important tasks in a safe setting. When the disease advances further, the patient goes to a memory care unit, where there is 24 hour supervision, specialized behavioral care, and constant medical attention. The final setting is hospice care, which is when the patient is at the end-of-life stage. This prioritizes emotional support, comfort, and pain relief. 
Source: https://health.clevelandclinic.org/long-term-care-for-alzheimers-patients

* **Biological mechanisms (anatomy, organ physiology, cell & molecular physiology):** In Alzheimer’s disease, the anatomical and organ physiology changes include, cerebral cortex atrophy, hippocampal degeneration, ventricular enlargement, and sulcal widening. The cellular physiology disruptions are synaptic dysfunction, where the communication junctions between neurons wither away, neuroinflammation, where pro-inflammatory cytokines are released, and mitochondrial energy failure, where there is metabolic dysfunction and a failure to produce sufficient ATP. The molecular physiology disruptions include Amyloid-beta plaque accumulation, which disrupt cell signaling, neurofibrillary tangling from Tau, which causes the cell’s transport system to collapse, and neurotransmitter deficiency, where a severe loss of cholinergic neurons causes a shortage of the vital neurotransmitter acetylcholine. 
Source: https://www.nia.nih.gov/health/alzheimers-causes-and-risk-factors/what-happens-brain-alzheimers-disease

* **Clinical Trials/next-gen therapies:** Amyloid beta-clearing drugs were approved by the FDA in recent years,  and they are called Leqembi and Kisunla. Studies have shown that they slow cognitive decline. Further amyloid plaque-clearing pills are currently being tested. For example, trontinemab has recently moved into Phase III clinical trials. The previously mentioned treatments are all infusions or injections, but there are pill-based treatments being investigated. For example, blarcamesine is in Phase II of the clinical trials.  There is continued research into non-pharmacological treatments for Alzheimer’s, which includes therapies such as focused ultrasound and brain stimulation. Incoming research supports the idea that lifestyle habits can help reduce the risk of developing Alzheimer’s. 
Source: https://www.brightfocus.org/resource/expanding-the-alzheimers-treatment-landscape-a-2026-forecast/

### Average MMSE Score by Age of Diagnosis
Null hypothesis: There is no statistical difference in the average MMSE score and the age of dementia diagnosis across various age groups (Under 70, 70-79, 80+).

Our p -value, 0.0034, is less than 0.05, therefore  we can reject the null hypothesis; there is a statistical difference in average MMSE scores and between different age of diagnosis groups.

Patients diagnosed with Alzheimer’s before age 70 had a lower average MMSE score than patients diagnosed at older ages, suggesting that earlier age of diagnosis was associated with lower cognitive function later in life. 


```python
#Bar Graph: Average MMSE Score by Age of Diagnosis

from patient_Mona import * #imports the patient file that has all necessary methods to gather and analyze patient data

#imports all necessary packages
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import statistics 
import pandas as pd
from sklearn.linear_model import LinearRegression #for the line of best fit


    #to be sure that you're still creating your patient objects from the .csv data file. 
Patient.instantiate_from_csv("/Users/monaabouassi/Library/Mobile Documents/com~apple~CloudDocs/BME 2315/MODULE 1/BME2315-Module-1/Metadata and Protein Data for Module 1 copy.csv")

    #makes two empty lists that will be populated by female and male patients, respectively
mmse_under_70 = []
mmse_70_79 = []
mmse_80_plus = []

 #populates these lists using the age of diagnosis filter
for patient in Patient.all_patients:

    if patient.age_of_diagnosis is not None and patient.last_MMSE_score is not None:

        if patient.age_of_diagnosis < 70:
            mmse_under_70.append(float(patient.last_MMSE_score))

        elif 70 <= patient.age_of_diagnosis < 80:
            mmse_70_79.append(float(patient.last_MMSE_score))

        else:
            mmse_80_plus.append(float(patient.last_MMSE_score))


    # Calculate means
mean_under_70 = statistics.mean(mmse_under_70)
mean_70_79 = statistics.mean(mmse_70_79)
mean_80_plus = statistics.mean(mmse_80_plus)

    # Calculate standard deviations
stdev_under_70 = statistics.stdev(mmse_under_70)
stdev_70_79 = statistics.stdev(mmse_70_79)
stdev_80_plus = statistics.stdev(mmse_80_plus)


    # Set up bar plot
diagnosis_groups = ["Under 70", "70-79", "80+"]

# Creates a list containing the average MMSE score for each age-of-diagnosis group
mean_mmse = [
    mean_under_70,
    mean_70_79,
    mean_80_plus
]

# Creates a list containing the standard deviation of MMSE scores for each age-of-diagnosis group
stdev_mmse = [
    stdev_under_70,
    stdev_70_79,
    stdev_80_plus
]

yerr = [np.zeros(len(mean_mmse)), stdev_mmse]

# BAR GRAPH
plt.figure(1)

plt.bar(
    diagnosis_groups,
    mean_mmse,
    yerr=yerr,
    capsize=10,
    color=["green", "blue", "purple"]
)

plt.title("Average MMSE Score by Age of Diagnosis")
plt.xlabel("Age of Diagnosis")
plt.ylabel("Average MMSE Score")


# ANOVA TEST CODE -----------------------------------------------

# Runs a one-way ANOVA test to compare the mean MMSE scores across the three age-of-diagnosis groups
f_stat, p_val = stats.f_oneway(
    mmse_under_70,   # MMSE scores for patients diagnosed before age 70
    mmse_70_79,      # MMSE scores for patients diagnosed between ages 70 and 79
    mmse_80_plus     # MMSE scores for patients diagnosed at age 80 or older
)

# Creates formatted text that summarizes the ANOVA results
anova_text = (
    f"ANOVA\n"               # Labels the statistical test as ANOVA
    f"F = {f_stat:.2f}\n"    # Displays the F-statistic rounded to 2 decimal places
    f"p = {p_val:.4f}"       # Displays the p-value rounded to 4 decimal places
)

#creates the details, such as location, color, fontsize, of the equation and data from the analyses.
plt.text(
    0.05, 0.95,
    anova_text,
    color= "red",
    transform=plt.gca().transAxes,
    fontsize=12,
    verticalalignment="top",
    horizontalalignment="left"
)

#prints the results from the statistical analysis
print("ANOVA statistical analysis: ")
print(f"F-statistic = {f_stat}")
print(f"p-value = {p_val}")





```

### pTau Level vs Last MMSE Score 
Null hypothesis: There is no linear relationship between the last MMSE score and pTau levels in the brain.

Our p-value is 0.1330, which is greater than 0.05. We fail to reject the null hypothesis, and we do not have sufficient evidence to conclude that pTAU level is significantly associated with the last MMSE score. Although the regression line shows a slight negative trend, the relationship is not statistically significant.


```python
#Scatter Plot: pTAU Level vs Last MMSE Score

from patient_Mona import * #imports the patient file that has all necessary methods to gather and analyze patient data

#imports all necessary packages
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import statistics 
import pandas as pd
from sklearn.linear_model import LinearRegression #for the line of best fit

    #makes two empty lists to hol the values
patient_ptau = []
patient_mmse = []

#adds all the available data from the dataset to the list, not including the ones that do not have data for that attribute
for patient in Patient.all_patients:
    if patient.pTAU != "n/a" and patient.last_MMSE_score is not None:
        patient_ptau.append(float(patient.pTAU))
        patient_mmse.append(float(patient.last_MMSE_score))

#convert lists to numpy array first, then do outlier test
patient_ptau = np.array(patient_ptau)
patient_mmse = np.array(patient_mmse)


#***********************
# OUTLIER TEST USING IQR
#***********************

# Find the first quartile (Q1) and third quartile (Q3) for pTAU values
Q1_ptau = np.percentile(patient_ptau, 25)
Q3_ptau = np.percentile(patient_ptau, 75)

# Calculate the interquartile range (IQR) for pTAU
IQR_ptau = Q3_ptau - Q1_ptau

# Calculate the lower and upper limits used to identify pTAU outliers
lower_ptau = Q1_ptau - 1.5 * IQR_ptau
upper_ptau = Q3_ptau + 1.5 * IQR_ptau


# Find the first quartile (Q1) and third quartile (Q3) for Last MMSE Scores
Q1_mmse = np.percentile(patient_mmse, 25)
Q3_mmse = np.percentile(patient_mmse, 75)

# Calculate the interquartile range (IQR) for Last MMSE Scores
IQR_mmse = Q3_mmse - Q1_mmse

# Calculate the lower and upper limits used to identify MMSE outliers
lower_mmse = Q1_mmse - 1.5 * IQR_mmse
upper_mmse = Q3_mmse + 1.5 * IQR_mmse


# Creates a Boolean array that marks data points as True if they are within
# the acceptable IQR range for BOTH pTAU and MMSE
non_outliers = (
    (patient_ptau >= lower_ptau) &    # pTAU value is above the lower bound
    (patient_ptau <= upper_ptau) &    # pTAU value is below the upper bound
    (patient_mmse >= lower_mmse) &    # MMSE score is above the lower bound
    (patient_mmse <= upper_mmse)      # MMSE score is below the upper bound
)


# Prints a heading before displaying any detected outliers
print("\nOutliers detected:")

# Pairs together the pTAU and MMSE values that were identified as outliers
for ptau, mmse in zip(
    patient_ptau[~non_outliers],      # Selects pTAU values marked as outliers
    patient_mmse[~non_outliers]       # Selects MMSE values marked as outliers
):
    # Prints the pTAU and MMSE values for each detected outlier
    print(f"pTAU = {ptau}, Last MMSE Score = {mmse}")


# Removes the outliers and keeps only the pTAU values within the accepted range
patient_ptau_filtered = patient_ptau[non_outliers]

# Removes the matching MMSE values so the x- and y-data stay paired correctly
patient_mmse_filtered = patient_mmse[non_outliers]

#now for the linear regression analysis using the filtered data
X = patient_mmse  #independent variable
y = patient_ptau   #dependent variable

#Linear Regression 
X = patient_ptau_filtered.reshape(-1, 1)
y = patient_mmse_filtered

model = LinearRegression()
model.fit(X, y)


#***********************
#p-value calculation
#***********************

# Calculates the Pearson correlation coefficient (r) and p-value
# between the x-values and y-values
r_value, p_value = stats.pearsonr(
    X.flatten(),  # Converts X from a 2D array into a 1D array for pearsonr()
    y             # y-values being compared with X
)

# Prints the Pearson correlation coefficient
# r shows the direction and strength of the linear relationship
print(f"Pearson correlation = {r_value}")

# Prints the p-value
# p < 0.05 generally indicates a statistically significant relationship
print(f"p-value = {p_value}")


#visualize these data on our scatter plot, by typing the following:
plt.scatter(X, y, color='blue')
plt.plot(X, model.predict(X), color="red")

plt.xlabel("pTAU Level")
plt.ylabel("Last MMSE Score")
plt.title("pTAU vs Last MMSE Score")


# Annotate equation
slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(X,y)

equation = (
    f"y = {slope:.2f}x + {intercept:.2f}\n"
    f"R^2 = {r2:.2f}\n"
    f"p = {p_value:.4f}"
)
plt.text(
    0.75, 0.95,
    equation,
    transform=plt.gca().transAxes,
    color="red",
    fontsize=12,
    verticalalignment="top",
    horizontalalignment="left"
)


```

```python
#Scatter Plot: Age of Diagnosis vs Age of Onset Symptoms

from patient_Mona import * #imports the patient file that has all necessary methods to gather and analyze patient data

#imports all necessary packages
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import statistics 
import pandas as pd
from sklearn.linear_model import LinearRegression #for the line of best fit

   
    #makes two empty lists to hol the values
patient_age_of_diagnosis = []
patient_age_of_onset_symptoms = []

for patient in Patient.all_patients: #if patient.age_of_diagnosis is not None and patient.age_of_onset_symptoms is not None:
    if patient.age_of_diagnosis is not None and patient.age_of_onset_symptoms is not None:
        patient_age_of_diagnosis.append(patient.age_of_diagnosis)
        patient_age_of_onset_symptoms.append(patient.age_of_onset_symptoms)


X = patient_age_of_diagnosis  #independent variable
y = patient_age_of_onset_symptoms   #dependent variable

#Linear Regression 
X = np.array(patient_age_of_diagnosis).reshape(-1,1)
y = np.array(patient_age_of_onset_symptoms) 

model = LinearRegression()
model.fit(X, y)

#***********************
# P-VALUE CALCULATION
#***********************

# Calculates the Pearson correlation coefficient (r) and p-value
# between the x-values and y-values
r_value, p_value = stats.pearsonr(
    X.flatten(),  # Converts X from a 2D array into a 1D array
    y             # y-values being compared with X
)

# Prints the Pearson correlation coefficient
# This shows the strength and direction of the linear relationship
print(f"Pearson correlation = {r_value}")

# Prints the p-value
# This helps determine whether the correlation is statistically significant
print(f"p-value = {p_value}")

#SCATTER PLOT
plt.figure(4)



    #visualize these data on our scatter plot, by typing the following:
plt.scatter(X, y, color='blue')
plt.plot(X, model.predict(X), color="red")

plt.xlabel('Age of Diagnosis')
plt.ylabel('Age of Onset Symptoms')
plt.title('Age of Diagnosis vs Age of Onset Symptoms')


# Annotate equation
slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(X,y)

equation = equation = (
    f"y = {slope:.2f}x + {intercept:.2f}\n"
    f"R^2 = {r2:.2f}\n"
    f"p = {p_value:.4f}"
)
plt.text(
    0.10, 0.95,
    equation,
    transform=plt.gca().transAxes,
    color="red",
    fontsize=12,
    verticalalignment="top",
    horizontalalignment="left"
)

```
